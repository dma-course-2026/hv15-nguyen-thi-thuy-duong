```python
# Cú pháp Python và chú thích

x = 5
if x > 0:
    print("Số dương")
else:
    print("Không dương")

# Chú thích một dòng
print("Python phân biệt chữ hoa và chữ thường")

"""
Chú thích nhiều dòng
có thể viết trong ba dấu nháy.
"""
print("Thụt lề quyết định khối lệnh")
```

```python
# Biến, gán nhiều biến, và kiểu dữ liệu

name = "Minh"
age = 20
price = 99.99
is_student = True

print(name, age, price, is_student)
print(type(name))
print(type(age))
print(type(price))
print(type(is_student))

a, b, c = 10, 20, 30
x = y = z = 5
print(a, b, c)
print(x, y, z)
```

```python
# Toán tử cơ bản

a = 10
b = 3

print("Số học:")
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)

print("\nSo sánh:")
print(a > b)
print(a == b)
print(a != b)

print("\nLogic và thành viên:")
print((a > 5) and (b < 5))
print((a < 5) or (b < 5))
print("a" in "apple")
print(2 in [1, 2, 3])
```

```python
# Nhập và xuất dữ liệu (dùng giá trị mẫu để thực hành nhanh)

name = "Minh"
age_text = "20"
age = int(age_text)

print("Xin chào,", name)
print("Sang năm bạn sẽ", age + 1, "tuổi.")
print("A", "B", "C", sep="-")
print("Xong", end="***")
print("Kết thúc")
```
