```python
# Bài 1:
Ket_qua = []
for i in range(2000, 3201):
    if i % 7 == 0 and i % 5 != 0:
        Ket_qua.append(i)
print(Ket_qua)
```

```python
# Bài 2:
n = int(input("Nhập số cần tính giai thừa: "))
giai_thua = 1
for i in range(1, n + 1):
    giai_thua *= i
print(giai_thua)
```

```python
# Bài 3:
n = int(input("Nhập số nguyên n: "))
ket_qua = {}    
for i in range(1, n + 1):
    ket_qua[i] = i * i
print(ket_qua)
```
