```python
# bài 5:
def so_sanh_chuoi(a,b):
    if a == b:
        return "Hai chuỗi bằng nhau"
    elif a > b:
        return "Chuỗi a lớn hơn chuỗi b"
    else:
        return "Chuỗi a nhỏ hơn chuỗi b"
a=input("nhập chuỗi a: ")
b=input("nhập chuỗi b: ")
print(so_sanh_chuoi(a, b))
```
