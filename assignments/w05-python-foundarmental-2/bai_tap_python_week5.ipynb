```python
# Bài 32 – Giải Hệ Phương Trình Bậc Nhất Hai Ẩn
a, b, c, d, m, n = map(float, input("Nhập các hệ số a, b, c, d, m, n: ").split())
print(f"-> Hệ số đã nhập: a = {a}, b = {b}, c = {c}, d = {d}, m = {m}, n = {n}") # in ra hệ số đã nhập
D = a * d - b * c
Dx = m * d - b * n
Dy = a * n - m * c
if D != 0:
    x = Dx / D
    y = Dy / D
    print(f"Hệ phương trình có nghiệm duy nhất: x = {x:.2f}, y = {y:.2f}")
elif D ==0 and Dx ==0 and Dy ==0:
    print ("Hệ phương trình có vô số nghiệm.")
else:
    print ("Hệ phương trình vô nghiệm")
```

```python
# Bài 33 – Điểm Thuộc Đường Tròn
import math
x0, y0 = map(float, input("nhập tọa độ tâm x0, y0:").split())
R= float(input("nhập bán kính R:"))
xa, ya = map(float, input("Nhập tọa độ điểm A (xa, ya):").split())
print(f"=> Các điểm đã nhập: x0={x0}, y0={y0}, xa={xa}, ya={ya}, R={R} ")
d = math.sqrt((xa -x0)**2 + (ya-y0)**2)
# Khai báo sai số epsilon
epsilon = 1e-9

# So sánh khoảng cách d với bán kính R theo gợi ý
if abs(d - R) < epsilon:
    print("Điểm A nằm trên đường tròn.")
elif d < R:
    print("Điểm A nằm trong đường tròn.")
else:
    print("Điểm A nằm ngoài đường tròn.")
```
