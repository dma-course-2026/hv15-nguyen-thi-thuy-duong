```python
#Bài 6 - Diện tích khu vườn còn lại
a = float(input("Nhập số a: "))
b = float(input("Nhập số b: "))
print (f"-> Các số đã nhập: a = {a}, b = {b}") # in ra các số đã nhập
s1 = a*b
pi = 3.14
s2 = pi*(a/2)**2
dien_tich_con_lại = s1 - s2
print(f"Diện tích còn lại là: {dien_tich_con_lại:.2f}")
```

```python
#Bài 7 - Điểm số
a, b, c = map(float, input("Nhập điểm số của 3 môn học: ").split())
diem_trung_binh = (a + b + c) / 3
print(f"-> Điểm 3 môn đã nhập: a = {a}, b = {b}, c = {c}") # in ra điểm 3 môn đã nhập
if diem_trung_binh >= 5.0:
    print("Điểm trung bình là:", diem_trung_binh, "- Học sinh đậu")
else:
    print("Điểm trung bình là:", diem_trung_binh, "- Học sinh trượt")
```

```python
# Bài 9 - Tính tiền điện
ho_va_ten = input("Nhập họ và tên chủ hộ: ")
chi_so_cu, chi_so_moi = map(float, input("Nhập chỉ số điện cũ và mới: ").split())
print(f"-> Chỉ số điện cũ: {chi_so_cu}, Chỉ số điện mới: {chi_so_moi}") # in ra chỉ số điện cũ và mới
luong_dien = chi_so_moi - chi_so_cu
if luong_dien <= 60:
    tien_dien = luong_dien * 5
elif luong_dien <=160:
    tien_dien = 60 * 5 + (luong_dien - 60) * 8
else:
    tien_dien = 60 * 5 + 100 * 8 + (luong_dien - 160) * 10

print(f"-> Họ và tên chủ hộ: {ho_va_ten}")
print(f"Tiêu thụ {luong_dien} kWh => số tiền điện phải trả là: {tien_dien:.2f} VND")
```

```python
# Bài 11 - Hình Vuông
a = float(input("Nhập độ dài cạnh a của hình vuông: "))
s = a * a
p = 4 * a
print(f"-> Độ dài cạnh a: {a}") # in ra độ dài cạnh a
print(f"Diện tích hình vuông: {s:.2f}") # in ra diện tích hình vuông
print(f"Chu vi hình vuông: {p:.2f}") # in ra chu vi hình vuông
```

```python
# Bài 12 - Hình chữ nhật
dai = float(input("Nhập độ dài cạnh dài của hình chữ nhật: "))
rong = float(input("Nhập độ dài cạnh rộng của hình chữ nhật: "))
print(f"chiều dài: {dai}, chiều rộng: {rong}") # in ra chiều dài và chiều rộng
s = dai * rong
p = 2 * (dai + rong)
print("Diện tích hình chữ nhật là:", s)
print("Chu vi hình chữ nhật là:", p)    
```

```python
# Bài 13 - Hình tròn
r = float(input("Nhập bán kính hình tròn: "))
print(f"bán kính: {r}") # in ra bán kính
s = 3.14 * r**2
p = 2 * 3.14 * r
print(f"Diện tích hình tròn là: {s:.2f}")
print(f"Chu vi hình tròn là: {p:.2f}")    
```

```python
# Bài 13 - Hình tam giác
a, b, c = map(float, input("Nhập độ dài 3 cạnh của hình tam giác: ").split())
print(f"-> Các cạnh đã nhập: a = {a}, b = {b}, c = {c}") # in ra các cạnh đã nhập
if a + b > c and a + c > b and b + c > a:
    p = (a + b + c) / 2
    s = (p * (p - a) * (p - b) * (p - c)) ** 0.5
    print(f"Diện tích hình tam giác là: {s:.2f}")
    print(f"Chu vi hình tam giác là: {a + b + c:.2f}")
```

```python
# Bài 15- Tam giác đều
a, b, c = map(float, input("Nhập độ dài 3 cạnh của hình tam giác đều: ").split())
print(f"-> Các cạnh đã nhập: a = {a}, b = {b}, c = {c}") # in ra các cạnh đã nhập
if a == b == c:
    print (f"-> Là tam giác đều")
else:
    print("-> Không phải là tam giác đều")   
```

```python
# Bài 17 - Tam giác vuông
a, b, c = map(float, input("Nhập độ dài 3 cạnh của hình tam giác vuông: ").split())
print(f"-> Các cạnh đã nhập: a = {a}, b = {b}, c = {c}")
if a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or b**2 + c**2 == a**2:
    print("-> Là tam giác vuông")
else:
    print("-> Không phải là tam giác vuông")
```

```python
# Bài 18 – Giải Phương Trình Bậc 1 (ax + b = 0)
a = float(input("Nhập hệ số a: "))
b = float(input("Nhập hệ số b: "))
print(f"-> Hệ số đã nhập: a = {a}, b = {b}") # in ra hệ số đã nhập
if a == 0:
    if b == 0:
        print("Phương trình có vô số nghiệm.")
    else:
        print("Phương trình vô nghiệm.")    
else:
    x = -b / a
    print(f"Phương trình có nghiệm duy nhất: x = {x:.2f}")       
```
