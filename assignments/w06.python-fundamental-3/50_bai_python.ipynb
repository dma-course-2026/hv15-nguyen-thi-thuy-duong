```python
# Bài 9. Tính tiền điện
def tinh_tien_dien():
    ho_ten=input ("Họ và tên: ")
    chi_so=input ("chỉ số tháng trước, chỉ số tháng này")
    chiso1, chiso2 = map(int, chi_so.split())
    luong_dien = chiso2 - chiso1
    if luong_dien <=60:
        tien = luong_dien*5
    elif luong_dien <=160:
        tien = (60*5)+ (luong_dien-60)*8
    else:
        tien =(60*5)+(100*8)+(luong_dien-160)*10
    return ho_ten, tien   
ten, so_tien = tinh_tien_dien()
print(f"Họ và tên: {ten}")
print(f"Tiền phải trả là {so_tien:.2f}")
            
```

```python
# Bài 32. Hệ phương trình bậc 1 hai ẩn
def ptbac1():
    a,b,m,c,d,n=map(float,input("nhập các số: ").split())
    D = a*d - b*c
    Dx = m*d -b*n
    Dy = a*n -c*m
    if D!=0:
        x= Dx/D
        y= Dy/D
        return f"Hệ pt có nghiệm duy nhất: x={x}, y={y}"
    else:
        if Dx==0 and Dy ==0:
            return "Hệ pt vô số nghiệm"
        else:
            return "Hệ pt vô nghiệm"
ptbac1()        
```

```python
# Bài 33. Điểm thuộc đường tròn
def diemthuocduongtron():
    import math
    R =float(input("nhập bán kính:"))
    x0= float(input("x0="))
    y0= float(input("y0="))
    xa= float(input("xa="))
    ya= float(input("ya="))
    d= math.sqrt((xa - x0)**2 + (ya - y0)**2)
    if d<R:
        return "A nằm trong đường tròn"
    elif d==R:
        return "A nằm trên đường tròn"
    else:
        return "A nằm ngoài đường tròn"
diemthuocduongtron()
```
