```python
# Bài 6. Diện tích khu vườn
def canh_hinh_chu_nhat(a,b, pi=3.14):
    s_vuong = a*b
    s_tron = pi * (a/2)**2
    return s_vuong - s_tron
print(canh_hinh_chu_nhat(10,20))    
```

```python
# Bài 7. Điểm trung bình
def nhap_ba_so_thuc(a,b,c):
    diem_trung_binh = round((a + b+ c)/3, 2)
    return diem_trung_binh, "Đậu" if diem_trung_binh>=5 else "Trượt"
print (nhap_ba_so_thuc(4.5,6.7,3.0))
```

```python
# Bài 8. Kiểm tra ký tự Alphabet
def nhap_mot_ky_tu(c):
    if  'a' <= c <= 'z' or 'A' <= c <= 'Z': 
      return f"{c} là ký tự alphabet"
    else:
      return "khong phai la ki tu alphabet"
print(nhap_mot_ky_tu('A'))    
```

```python
# Bài 9. Tính tiền điện
def tinh_tien_dien():
    ho_ten=input ("Họ và tên: ")
    chi_so=input ("chỉ số tháng trước, chỉ số tháng này")
    chiso1, chiso2 = map(int, chi_so.split())
    luong_dien = chiso2 - chiso1
    if luong_dien <=60:
        tien = luong_dien*5
    elif luong_dien <=160:
        tien = (60*5)+ (luong_dien-60)*8
    else:
        tien =(60*5)+(100*8)+(luong_dien-160)*10
    return ho_ten, tien   
ten, so_tien = tinh_tien_dien()
print(f"Họ và tên: {ten}")
print(f"Tiền phải trả là {so_tien:.2f}")
            
```

```python
# Bài 16. Tam giác cân
def tamgiaccan():
    a, b, c =map(float,input("nhập 3 cạnh của tam giác:").split())
    print(f" Các cạnh đã nhập: a={a}, b={b}, c={c}")
    if a==b or b==c or a==c:
        return True
    else:
        return False
ket_qua= tamgiaccan() 
if ket_qua==True:
    print("Tam giác cân")
else:
    print("không phải tam giác cân")

```
